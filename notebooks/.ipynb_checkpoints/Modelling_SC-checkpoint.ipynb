```python

def run_classification_models(
    df,
    feature_cols,
    target_col='log_acres',
    test_size=0.2,
    random_state=42
):
    data = df[feature_cols + [target_col]].copy()
    data = data.dropna(subset=[target_col])

    # counts, edges = np.histogram(df["log_acres"], bins=8)

    # data["fire_size_class"] = pd.cut(
    #     data["log_acres"],
    #     bins=edges,
    #     include_lowest=True
    # )

    # data["acres"] = np.expm1(data[target_col])

    # data["fire_size_class"] = pd.cut(
    #     data["acres"],
    #     bins=[0, 10, 25, 50, 100, 250, 500, 1000, np.inf],
    #     labels=[
    #         "0-10",
    #         "10-25",
    #         "25-50",
    #         "50-100",
    #         "100-250",
    #         "250-500",
    #         "500-1000",
    #         ">1000"
    #     ],
    #     include_lowest=True
    # )

    data['fire_size_class'] = pd.qcut(
    data[target_col],
    q=3,
    labels=['Very Low', 'Medium', 'High']
    )

    
    # data['fire_size_class'] = pd.qcut(
    #     data[target_col],
    #     q=5,
    #     labels=['Very Low', 'Low-Med', 'Medium', 'High','Very High']
    # )

    X = data[feature_cols]
    y = data['fire_size_class']

    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=test_size,
        random_state=random_state,
        stratify=y_encoded
    )

    models = {
        "Logistic Regression": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(
                max_iter=5000,
                random_state=random_state,
                class_weight="balanced"
            ))
        ]),

        "Random Forest": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestClassifier(
                n_estimators=500,
                max_depth=5,
                min_samples_leaf=25,
                min_samples_split=50,
                max_features=0.4,
                random_state=random_state,
                n_jobs=-1,
                class_weight="balanced"
            ))
        ]),

        "Extra Trees": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", ExtraTreesClassifier(
                n_estimators=500,
                max_depth=5,
                min_samples_leaf=20,
                min_samples_split=40,
                max_features=0.4,
                random_state=random_state,
                n_jobs=-1,
                class_weight="balanced"
            ))
        ]),

        "HistGradientBoosting": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", HistGradientBoostingClassifier(
                random_state=random_state,
                max_iter=300,
                max_depth=4,
                min_samples_leaf=35,
                l2_regularization=10.0
            ))
        ]),


        "Neural Network": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPClassifier(
                hidden_layer_sizes=(64, 32),
                activation="relu",
                solver="adam",
                alpha=0.01,
                learning_rate_init=0.001,
                max_iter=1000,
                early_stopping=True,
                validation_fraction=0.15,
                n_iter_no_change=25,
                random_state=random_state
            ))
        ])
    }

    try:
        from lightgbm import LGBMClassifier

        models["LightGBM"] = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", LGBMClassifier(
                n_estimators=400,
                learning_rate=0.02,
                max_depth=4,
                num_leaves=15,
                min_child_samples=35,
                subsample=0.7,
                colsample_bytree=0.7,
                reg_alpha=10.0,
                reg_lambda=15.0,
                random_state=random_state,
                n_jobs=-1,
                verbose=-1
            ))
        ])
    except ImportError:
        print("LightGBM is not installed. Skipping LightGBM.")

    try:
        from xgboost import XGBClassifier

        models["XGBoost"] = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", XGBClassifier(
                n_estimators=400,
                learning_rate=0.03,
                max_depth=3,
                subsample=0.7,
                colsample_bytree=0.7,
                reg_alpha=10.0,
                reg_lambda=15.0,
                eval_metric="mlogloss",
                random_state=random_state,
                n_jobs=-1
            ))
        ])
    except ImportError:
        print("XGBoost is not installed. Skipping XGBoost.")

    results = []
    fitted_models = {}

    for name, model in models.items():
        model.fit(X_train, y_train)

        y_train_pred = model.predict(X_train)
        y_test_pred = model.predict(X_test)

        cm = confusion_matrix(y_test, y_test_pred)

        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm,
            display_labels=label_encoder.classes_
        )
        
        # fig, ax = plt.subplots(figsize=(5, 5))
        # disp.plot(ax=ax, cmap="Blues", colorbar=False)
        # plt.title(f"{name} - Confusion Matrix")
        # plt.tight_layout()
        # plt.show()

        train_acc = accuracy_score(y_train, y_train_pred)
        test_acc = accuracy_score(y_test, y_test_pred)

        results.append({
            "Model": name,

            "Train Accuracy": train_acc,
            "Test Accuracy": test_acc,

            "Train Precision": precision_score(
                y_train,
                y_train_pred,
                average="weighted",
                zero_division=0
            ),
            "Test Precision": precision_score(
                y_test,
                y_test_pred,
                average="weighted",
                zero_division=0
            ),

            "Train Recall": recall_score(
                y_train,
                y_train_pred,
                average="weighted",
                zero_division=0
            ),
            "Test Recall": recall_score(
                y_test,
                y_test_pred,
                average="weighted",
                zero_division=0
            ),

            "Train F1": f1_score(
                y_train,
                y_train_pred,
                average="weighted",
                zero_division=0
            ),
            "Test F1": f1_score(
                y_test,
                y_test_pred,
                average="weighted",
                zero_division=0
            ),

            "Overfit Gap": train_acc - test_acc
        })

        fitted_models[name] = model

    results_df = (
        pd.DataFrame(results)
        .sort_values("Test F1", ascending=False)
        .reset_index(drop=True)
    )

    return results_df, fitted_models, label_encoder
```

```python

def run_regression_models(df, feature_cols, target_col='log_acres', test_size=0.2, random_state=42):
    
    data = df[feature_cols + [target_col]].copy()
    # data = data.dropna(subset=[target_col])
    data = data.dropna()

    X = data[feature_cols]
    y = data[target_col]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=test_size,
        random_state=random_state
    )

    models = {
        "Linear Regression": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LinearRegression())
        ]),

        "Elastic Net": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", ElasticNetCV(cv=5, random_state=random_state, max_iter=10000))
        ]),


        "Random Forest": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestRegressor(
                n_estimators=500,
                max_depth=5,                 
                min_samples_leaf=25,        
                min_samples_split=50,        
                max_features=0.4,            
                random_state=random_state,
                n_jobs=-1
            ))
        ]),

        "Extra Trees": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", ExtraTreesRegressor(
                n_estimators=500,
                max_depth=5,
                min_samples_leaf=20,
                min_samples_split=40,
                max_features=0.4,
                random_state=random_state,
                n_jobs=-1
            ))
        ]),


        "HistGradientBoosting": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", HistGradientBoostingRegressor(
                random_state=random_state,
                max_iter=150,                
                max_depth=3,                 
                min_samples_leaf=50,         
                l2_regularization=20.0      ,
                learning_rate=0.03,
                max_leaf_nodes=15,
                early_stopping=True
            ))
        ]),

        "Support Vector Regressor": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", SVR(
                kernel="rbf",
                C=10,
                epsilon=0.1,
                gamma="scale"
            ))
        ]),

        "Gaussian Process": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", GaussianProcessRegressor(
                kernel=(
                    ConstantKernel(1.0, (1e-2, 1e2))
                    * RBF(length_scale=1.0, length_scale_bounds=(1e-2, 1e2))
                    + WhiteKernel(noise_level=1.0, noise_level_bounds=(1e-5, 1e1))
                ),
                alpha=1e-2,
                normalize_y=True,
                n_restarts_optimizer=2,
                random_state=random_state
            ))
        ]),

           "Neural Network": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPRegressor(
                hidden_layer_sizes=(64, 32),
                activation="relu",
                solver="adam",
                alpha=0.01,
                learning_rate_init=0.001,
                max_iter=1000,
                early_stopping=True,
                validation_fraction=0.15,
                n_iter_no_change=25,
                random_state=random_state
            ))
        ]),
    }

 

    try:
        from lightgbm import LGBMRegressor
        
        models["LightGBM"] = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", LGBMRegressor(
                n_estimators=200,
                learning_rate=0.01,          
                max_depth=3,                 
                num_leaves=15,               
                min_child_samples=50,        
                subsample=0.75,               
                colsample_bytree=0.75,        
                reg_alpha=5.0,              
                reg_lambda=20.0,             
                random_state=random_state,
                alpha= 0.90,
                n_jobs=-1,
                verbose=-1
            ))
        ])
    except ImportError:
        print("LightGBM is not installed. Skipping LightGBM.")


    try:
        from xgboost import XGBRegressor

        models["XGBoost"] = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", XGBRegressor(
                n_estimators=120,
                learning_rate=0.02,          
                max_depth=7,
                min_child_weight=20,
                subsample=0.8,          
                colsample_bytree=0.7,
                gamma=2.0,
                reg_alpha=0,              
                reg_lambda=10,             
                objective="reg:squarederror",
                random_state=random_state,
                n_jobs=-1
            ))
        ])
        

    except ImportError:
        print("XGBoost is not installed. Skipping XGBoost.")


    try:
        from catboost import CatBoostRegressor
    
        models["CatBoost"] = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", CatBoostRegressor(
                iterations=500,
                learning_rate=0.03,
                depth=4,
                l2_leaf_reg=15,
                loss_function="RMSE",
                random_seed=random_state,
                verbose=False
            ))
        ])

    except ImportError:
        print("CatBoost is not installed. Skipping CatBoost.")


    results = []

    for name, model in models.items():
        model.fit(X_train, y_train)

        # print importance
        if name == 'XGBoost--':
            best_model = models["XGBoost"]
        
            xgb_model = best_model.named_steps["model"]
            
            importances = pd.DataFrame({
                "feature": feature_cols,
                "importance": xgb_model.feature_importances_
            }).sort_values("importance", ascending=False)
            
            print(importances.head(30))

        y_train_pred = model.predict(X_train)
        y_test_pred = model.predict(X_test)


        results.append({
            "Model": name,

            "Train RMSE": np.sqrt(mean_squared_error(y_train, y_train_pred)),
            "Test RMSE": np.sqrt(mean_squared_error(y_test, y_test_pred)),

            "Train MAE": mean_absolute_error(y_train, y_train_pred),
            "Test MAE": mean_absolute_error(y_test, y_test_pred),

            "Train R2": r2_score(y_train, y_train_pred),
            "Test R2": r2_score(y_test, y_test_pred),

            "Overfit Gap": r2_score(y_train, y_train_pred) - r2_score(y_test, y_test_pred)
        })

    results_df = pd.DataFrame(results).sort_values("Test R2", ascending=False)

    

    

    return results_df
```

### Feature engineering

```python
scaler = StandardScaler()
df['t_log_spatial_local_urban_population'] = np.log1p(df['spatial_local_urban_population'])
df['std_spatial_distance_to_road_km']  = scaler.fit_transform(df[['spatial_distance_to_road_km']])
df['std_spatial_distance_to_city_km']  = scaler.fit_transform(df[['spatial_distance_to_city_km']])
df['std_spatial_distance_to_coast_km']  = scaler.fit_transform(df[['spatial_distance_to_coast_km']])
df['std_spatial_distance_to_powerline_km']  = scaler.fit_transform(df[['spatial_distance_to_powerline_km']])
df['std_spatial_local_urban_population']  = scaler.fit_transform(df[['spatial_local_urban_population']])
df['std_weather_erc_mean_2w']  = scaler.fit_transform(df[['weather_erc_mean_2w']])
df['std_weather_vpd_mean_2w']  = scaler.fit_transform(df[['weather_vpd_mean_2w']])
df['std_drought_dsci_current']  = scaler.fit_transform(df[['drought_dsci_current']])

# df.drop(columns=['t_log_spatial_local_urban_population',
#                  'std_spatial_distance_to_road_km',
#                  'std_spatial_distance_to_city_km',
#                  'std_spatial_distance_to_coast_km',
#                  'std_spatial_distance_to_powerline_km',
#                  'std_spatial_local_urban_population'
#                 ]
#                  ,inplace=True)
```

```python

# convert object to date
df['fire_incident_date_created'] = pd.to_datetime(df['fire_incident_date_created'])
df['fire_incident_date_last_update'] = pd.to_datetime(df['fire_incident_date_last_update'])

# calc days burnt
df['days_burnt'] = (df['fire_incident_date_last_update'] - df['fire_incident_date_created']).dt.days

# handle null values
median_days_burnt = df['days_burnt'].median()
df['days_burnt'] = df['days_burnt'].fillna(median_days_burnt)

# to sort back in original order
df['_orig_order'] = range(len(df))

# sort values
df = df.sort_values(['fire_incident_county', 'fire_incident_date_created'])

# calc cumulative count
df['feat_previous_incident_count_by_county'] = (
    df.groupby('fire_incident_county')
      .cumcount()
)

# historical mean log acres
df['_cum_sum_acres'] = df.groupby('fire_incident_county')['log_acres'].cumsum()
df['_past_sum_acres'] = df['_cum_sum_acres'] - df['log_acres']
df['feat_county_historical_mean_log_acres'] = df['_past_sum_acres'] / df['feat_previous_incident_count_by_county']

# historical mean days count burnt
df['_cum_sum_days'] = df.groupby('fire_incident_county')['days_burnt'].cumsum()
df['_past_sum_days'] = df['_cum_sum_days'] - df['days_burnt']
df['feat_county_historical_mean_days_burnt'] = df['_past_sum_days'] / df['feat_previous_incident_count_by_county']

global_mean_acres = df['log_acres'].mean()
global_mean_days = df['days_burnt'].mean()

df['feat_county_historical_mean_log_acres'] = df['feat_county_historical_mean_log_acres'].fillna(global_mean_acres)
df['feat_county_historical_mean_days_burnt'] = df['feat_county_historical_mean_days_burnt'].fillna(global_mean_days)

# restore original row and drop processing columns
df = df.sort_values('_orig_order').drop(columns=[
    '_orig_order', 
    '_cum_sum_acres', 
    '_past_sum_acres',
    '_cum_sum_days', 
    '_past_sum_days'
])
```

```python
df['feat_topography_elevation_x_weather_erc_mean_2w'] = df['topography_elevation'] * df['weather_erc_mean_2w']
```

```python
df['feat_hot_dry_windy_proxy'] = df['weather_tmmx_mean_8w'] * (100 - df['weather_rmin_mean_8w']) * df['weather_vs_mean_8w']
```

```python
# create original order and sort values
df['_orig_order'] = range(len(df))
df = df.sort_values(['vegetation_fuel_type', 'fire_incident_date_created'])

# 1. Running count of fires inside this fuel type
df['_veg_count'] = df.groupby('vegetation_fuel_type').cumcount()

# 2. Running sum of fire sizes in this fuel type
df['_veg_cum_sum'] = df.groupby('vegetation_fuel_type')['log_acres'].cumsum()

# 3. Exclude current row to prevent target leakage
df['_veg_past_sum'] = df['_veg_cum_sum'] - df['log_acres']
df['feat_fuel_type_historical_mean_log_acres'] = df['_veg_past_sum'] / df['_veg_count']

# 4. Fill initial instances with the global average
df['feat_fuel_type_historical_mean_log_acres'] = df['feat_fuel_type_historical_mean_log_acres'].fillna(df['log_acres'].mean())

# Restore original order
df = df.sort_values('_orig_order').drop(columns=['_orig_order', '_veg_count', '_veg_cum_sum', '_veg_past_sum'])
```

```python
df['feat_offshore_wind_delivery_sin_8w'] = ((df['weather_th_sin_8w'] >= 22.5) & (df['weather_th_sin_8w'] <= 67.5)).astype(int) * df['weather_vs_mean_8w']
```

```python
df['feat_offshore_wind_delivery_cos_8w'] = ((df['weather_th_cos_8w'] >= 22.5) & (df['weather_th_cos_8w'] <= 67.5)).astype(int) * df['weather_vs_mean_8w']
```

```python
df["feat_temp_range_2w"] = (
    df["weather_tmmx_mean_2w"] -
    df["weather_tmmn_mean_2w"]
)

df["feat_humidity_range_2w"] = (
    df["weather_rmax_mean_2w"] -
    df["weather_rmin_mean_2w"]
)

df["feat_fm_difference"] = (
    df["weather_fm1000_mean_2w"] -
    df["weather_fm100_mean_2w"]
)


df["feat_erc_change"] = (
    df["weather_erc_mean_2w"] -
    df["weather_erc_mean_8w"]
)

df["feat_vpd_change"] = (
    df["weather_vpd_mean_2w"] -
    df["weather_vpd_mean_8w"]
)

df["feat_temp_change"] = (
    df["weather_tmmx_mean_2w"] -
    df["weather_tmmx_mean_8w"]
)

df["feat_precip_change"] = (
    df["weather_pr_sum_2w"] -
    df["weather_pr_sum_8w"]
)

df["feat_hot_dry_windy"] = (
    df["weather_tmmx_mean_2w"] *
    df["weather_vpd_mean_2w"] *
    df["weather_vs_mean_2w"]
)

df["feat_slope_x_wind"] = (
    df["topography_slope"] *
    df["weather_vs_mean_2w"]
)

df["feat_slope_x_erc"] = (
    df["topography_slope"] *
    df["weather_erc_mean_2w"]
)

df["feat_elevation_x_temperature"] = (
    df["topography_elevation"] *
    df["weather_tmmx_mean_2w"]
)

df["feat_erc_x_fm100"] = (
    df["weather_erc_mean_2w"] *
    df["weather_fm100_mean_2w"]
)

df["feat_vpd_x_fm100"] = (
    df["weather_vpd_mean_2w"] *
    df["weather_fm100_mean_2w"]
)

df["feat_dsci_x_erc"] = (
    df["drought_dsci_current"] *
    df["weather_erc_mean_2w"]
)

df["feat_dsci_x_vpd"] = (
    df["drought_dsci_current"] *
    df["weather_vpd_mean_2w"]
)

df["feat_pdsi_x_precip"] = (
    df["drought_pdsi_current"] *
    df["weather_pr_sum_2w"]
)

df["feat_population_x_road"] = (
    df["t_log_spatial_local_urban_population"] /
    (df["spatial_distance_to_road_km"] + 1)
)

df["feat_population_x_powerline"] = (
    df["t_log_spatial_local_urban_population"] /
    (df["spatial_distance_to_powerline_km"] + 1)
)

df["feat_accessibility"] = (
    df["spatial_distance_to_road_km"] +
    df["spatial_distance_to_city_km"]
)

df["feat_remote_index"] = (
    df["spatial_distance_to_city_km"] +
    df["spatial_distance_to_powerline_km"] +
    df["spatial_distance_to_road_km"]
)

df["feat_fuel_percent"] = (
    df["vegetation_percent"] *
    df["weather_erc_mean_2w"]
)

df["feat_fuel_drought"] = (
    df["vegetation_percent"] *
    df["drought_dsci_current"]
)

df["feat_offshore_wind"] = (
    df["weather_th_cos_2w"] *
    df["weather_vs_mean_2w"]
)

df["feat_crosswind"] = (
    df["weather_th_sin_2w"] *
    df["weather_vs_mean_2w"]
)

df["feat_history_x_erc"] = (
    df["feat_previous_incident_count_by_county"] *
    df["weather_erc_mean_2w"]
)

df["feat_history_x_population"] = (
    df["feat_previous_incident_count_by_county"] *
    df["t_log_spatial_local_urban_population"]
)

df["fire_incident_date_created"] = pd.to_datetime(
    df["fire_incident_date_created"]
)

df["month"] = df.fire_incident_date_created.dt.month

df["day_of_year"] = (
    df.fire_incident_date_created.dt.dayofyear
)

df["feat_month_sin"] = np.sin(
    2*np.pi*df.month/12
)

df["feat_month_cos"] = np.cos(
    2*np.pi*df.month/12
)

df["feat_day_sin"] = np.sin(
    2*np.pi*df.day_of_year/365
)

df["feat_day_cos"] = np.cos(
    2*np.pi*df.day_of_year/365
)

df["feat_log_erc"] = np.log1p(df.weather_erc_mean_2w)

df["feat_log_vpd"] = np.log1p(df.weather_vpd_mean_2w)

df["feat_log_precip"] = np.log1p(df.weather_pr_sum_2w)

df["feat_erc_squared"] = df.weather_erc_mean_2w**2

df["feat_temperature_squared"] = (
    df.weather_tmmx_mean_2w**2
)

```

```python
def tune_xgboost_regressor(
    df,
    feature_cols,
    target_col="log_acres",
    test_size=0.2,
    random_state=42,
    n_iter=50,
    cv=5
):


    data = df[feature_cols + [target_col]].copy()
    data = data.dropna(subset=[target_col])

    X = data[feature_cols]
    y = data[target_col]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=test_size,
        random_state=random_state
    )

    pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", XGBRegressor(
            objective="reg:squarederror",
            random_state=random_state,
            n_jobs=-1
        ))
    ])

    param_dist = {
    "model__n_estimators": [100, 200, 300, 500, 700],
    "model__learning_rate": [0.01, 0.02, 0.03, 0.05],
    "model__max_depth": [2, 3, 4, 5],
    "model__min_child_weight": [5, 10, 20, 40, 60],
    "model__subsample": [0.6, 0.7, 0.8],
    "model__colsample_bytree": [0.6, 0.7, 0.8],
    "model__gamma": [0, 1, 2, 5],
    "model__reg_alpha": [0, 1, 5, 10, 20],
    "model__reg_lambda": [1, 5, 10, 20, 40]
}
    search = RandomizedSearchCV(
        estimator=pipe,
        param_distributions=param_dist,
        n_iter=n_iter,
        scoring="r2",
        cv=cv,
        random_state=random_state,
        verbose=2,
        n_jobs=-1
    )

    search.fit(X_train, y_train)

    best_model = search.best_estimator_

    y_train_pred = best_model.predict(X_train)
    y_test_pred = best_model.predict(X_test)

    results_df = pd.DataFrame([{
        "Model": "Tuned XGBoost",
        "Best CV R2": search.best_score_,
        "Train RMSE": np.sqrt(mean_squared_error(y_train, y_train_pred)),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, y_test_pred)),
        "Train MAE": mean_absolute_error(y_train, y_train_pred),
        "Test MAE": mean_absolute_error(y_test, y_test_pred),
        "Train R2": r2_score(y_train, y_train_pred),
        "Test R2": r2_score(y_test, y_test_pred),
        "Overfit Gap": r2_score(y_train, y_train_pred) - r2_score(y_test, y_test_pred),
        "Best Params": search.best_params_
    }])

    return results_df, best_model, search
```

```python
def tune_lightgbm_regressor(
    df,
    feature_cols,
    target_col="log_acres",
    test_size=0.2,
    random_state=42,
    n_iter=100,
    cv=5
):
    import numpy as np
    import pandas as pd

    from sklearn.model_selection import train_test_split, RandomizedSearchCV
    from sklearn.pipeline import Pipeline
    from sklearn.impute import SimpleImputer
    from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

    from lightgbm import LGBMRegressor

    data = df[feature_cols + [target_col]].copy()
    data = data.dropna(subset=[target_col])

    X = data[feature_cols]
    y = data[target_col]

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_size,
        random_state=random_state
    )

    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", LGBMRegressor(
            objective="regression",
            random_state=random_state,
            n_jobs=-1,
            verbose=-1
        ))
    ])

    param_dist = {
        "model__n_estimators": [200, 300, 500, 800, 1000],
        "model__learning_rate": [0.005, 0.01, 0.02, 0.03, 0.05],
        "model__max_depth": [2, 3, 4, 5, -1],
        "model__num_leaves": [7, 15, 31, 63],
        "model__min_child_samples": [20, 35, 50, 75, 100],
        "model__subsample": [0.6, 0.7, 0.8, 0.9, 1.0],
        "model__colsample_bytree": [0.5, 0.6, 0.7, 0.8, 0.9, 1.0],
        "model__reg_alpha": [0, 0.1, 0.5, 1, 2, 5, 10],
        "model__reg_lambda": [1, 2, 5, 10, 15, 20, 30]
    }

    search = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions=param_dist,
        n_iter=n_iter,
        scoring="r2",
        cv=cv,
        random_state=random_state,
        verbose=2,
        n_jobs=-1,
        return_train_score=True
    )

    search.fit(X_train, y_train)

    best_model = search.best_estimator_

    train_pred = best_model.predict(X_train)
    test_pred = best_model.predict(X_test)

    results = pd.DataFrame([{
        "Model": "Tuned LightGBM",
        "Best Params": str(search.best_params_),
        "CV Score": search.best_score_,
        "Train RMSE": np.sqrt(mean_squared_error(y_train, train_pred)),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, test_pred)),
        "Train MAE": mean_absolute_error(y_train, train_pred),
        "Test MAE": mean_absolute_error(y_test, test_pred),
        "Train R2": r2_score(y_train, train_pred),
        "Test R2": r2_score(y_test, test_pred),
        "Overfit Gap": r2_score(y_train, train_pred) - r2_score(y_test, test_pred)
    }])

    return results, best_model, search
```

```python
xgb_results_df, best_xgb_model, xgb_search = tune_xgboost_regressor(
    df=df,
    feature_cols=feature_cols,
    target_col="log_acres",
    n_iter=50,
    cv=5
)

xgb_results_df
```
